```python
import numpy as np, jax.numpy as jnp, pandas as pd
```

```python
from jax import grad
```

```python
def ejemplo(x: np.ndarray):
    return sum((xi ** (2*i+2) for i, xi in enumerate(x)))
```

```python
punto = jnp.array([1.0,0.0, 1.0])
grad(ejemplo)(punto)
```
